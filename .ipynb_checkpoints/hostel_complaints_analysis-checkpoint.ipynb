# Hostel Complaints Analysis
Exploring the `hostel_complaints_800_final (1).csv` dataset.

```
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('hostel_complaints_800_final (1).csv')
print(f'Dataset shape: {df.shape}')
df.head()
```

```
# Basic info
df.info()
```

```
# Summary statistics
df.describe(include='all')
```

```
# Check for missing values
df.isnull().sum()
```

```
# Value counts for categorical columns
for col in ['Category', 'Priority', 'Status', 'Block', 'Floor']:
    print(f'\n--- {col} ---')
    print(df[col].value_counts())
```

```
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Hostel Complaints Overview', fontsize=16)

df['Category'].value_counts().plot(kind='bar', ax=axes[0, 0], color='steelblue')
axes[0, 0].set_title('Complaints by Category')
axes[0, 0].tick_params(axis='x', rotation=45)

df['Priority'].value_counts().plot(kind='pie', ax=axes[0, 1], autopct='%1.1f%%')
axes[0, 1].set_title('Priority Distribution')

df['Status'].value_counts().plot(kind='bar', ax=axes[1, 0], color='coral')
axes[1, 0].set_title('Status Distribution')
axes[1, 0].tick_params(axis='x', rotation=45)

df['Block'].value_counts().plot(kind='bar', ax=axes[1, 1], color='mediumseagreen')
axes[1, 1].set_title('Complaints by Block')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```

# ════════════════════════════════════════════════════════════════════════
#  SVM MODEL EVALUATION
# ════════════════════════════════════════════════════════════════════════
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, confusion_matrix
)
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import numpy as np

# ── Step 1: Get class names from LabelEncoder ──────────────────────────
# label_encoder was fitted during training; its .classes_ attribute holds
# the original string labels (e.g. ['High', 'Low', 'Medium']).
class_names = label_encoder.classes_        # e.g. array(['High', 'Low', 'Medium'])
print(f"Class names (from LabelEncoder): {class_names.tolist()}")

# ── Step 2: Predict on the test set (NO retraining) ───────────────────
# svm_model and X_test already exist from the training section above.
y_pred = svm_model.predict(X_test)

# ── Step 3: Core metrics ───────────────────────────────────────────────
accuracy  = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
recall    = recall_score(y_test, y_pred, average='weighted', zero_division=0)
f1        = f1_score(y_test, y_pred, average='weighted', zero_division=0)

print("\n" + "="*55)
print("  SVM MODEL — TEST SET METRICS")
print("="*55)
print(f"  Accuracy  : {accuracy:.4f}  ({accuracy*100:.2f}%)")
print(f"  Precision : {precision:.4f}")
print(f"  Recall    : {recall:.4f}")
print(f"  F1-Score  : {f1:.4f}")
print("="*55)

# ── Step 4: Training vs Testing accuracy (overfitting check) ──────────
# Training accuracy should be close to testing accuracy for a well-
# generalised model. A large gap signals overfitting.
train_accuracy = accuracy_score(y_train, svm_model.predict(X_train))
test_accuracy  = accuracy_score(y_test,  y_pred)
gap            = train_accuracy - test_accuracy

print(f"\n  Overfitting Check:")
print(f"    Training Accuracy : {train_accuracy:.4f}  ({train_accuracy*100:.2f}%)")
print(f"    Testing  Accuracy : {test_accuracy:.4f}  ({test_accuracy*100:.2f}%)")
print(f"    Gap (Train-Test)  : {gap:.4f}")
if gap < 0.05:
    print("    ✅ Model generalises well — no significant overfitting.")
elif gap < 0.10:
    print("    ⚠️  Slight overfitting detected.")
else:
    print("    ❌ Significant overfitting — consider regularisation or more data.")

# ── Step 5: Full Classification Report ────────────────────────────────
# Shows per-class precision, recall, F1, and support using original labels.
print("\n" + "="*55)
print("  CLASSIFICATION REPORT")
print("="*55)
print(classification_report(
    y_test, y_pred,
    target_names=class_names,   # original string labels
    zero_division=0
))

# ── Step 6: Confusion Matrix ───────────────────────────────────────────
# Rows = true labels, Columns = predicted labels.
# Values on the diagonal are correct predictions.
cm = confusion_matrix(y_test, y_pred)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle("SVM Model Evaluation", fontsize=15, fontweight="bold", y=1.02)

# ── 6a. Raw count heatmap ──────────────────────────────────────────────
sns.heatmap(
    cm,
    annot=True,
    fmt="d",                        # integer counts
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names,
    linewidths=0.5,
    linecolor="lightgrey",
    ax=axes[0],
    cbar_kws={"label": "Count"}
)
axes[0].set_title("Confusion Matrix (Counts)", fontsize=12)
axes[0].set_xlabel("Predicted Label", fontsize=11)
axes[0].set_ylabel("True Label", fontsize=11)
axes[0].tick_params(axis="x", rotation=30)
axes[0].tick_params(axis="y", rotation=0)

# ── 6b. Normalised (percentage) heatmap ───────────────────────────────
cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)
sns.heatmap(
    cm_norm,
    annot=True,
    fmt=".2f",                      # two decimal places
    cmap="Greens",
    xticklabels=class_names,
    yticklabels=class_names,
    linewidths=0.5,
    linecolor="lightgrey",
    ax=axes[1],
    vmin=0, vmax=1,
    cbar_kws={"label": "Proportion"}
)
axes[1].set_title("Confusion Matrix (Normalised)", fontsize=12)
axes[1].set_xlabel("Predicted Label", fontsize=11)
axes[1].set_ylabel("True Label", fontsize=11)
axes[1].tick_params(axis="x", rotation=30)
axes[1].tick_params(axis="y", rotation=0)

plt.tight_layout()
plt.show()

# ── Step 7: Per-class metric bar chart ────────────────────────────────
from sklearn.metrics import precision_recall_fscore_support

precisions, recalls, f1s, supports = precision_recall_fscore_support(
    y_test, y_pred, labels=np.arange(len(class_names)), zero_division=0
)

x = np.arange(len(class_names))
width = 0.25

fig2, ax = plt.subplots(figsize=(9, 5))
bars_p = ax.bar(x - width, precisions, width, label="Precision", color="#4C72B0", alpha=0.85)
bars_r = ax.bar(x,         recalls,   width, label="Recall",    color="#55A868", alpha=0.85)
bars_f = ax.bar(x + width, f1s,       width, label="F1-Score",  color="#C44E52", alpha=0.85)

# Annotate bar values
for bars in [bars_p, bars_r, bars_f]:
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width() / 2, h + 0.01,
                f"{h:.2f}", ha="center", va="bottom", fontsize=8)

ax.set_xticks(x)
ax.set_xticklabels(class_names, fontsize=11)
ax.set_ylim(0, 1.15)
ax.set_ylabel("Score", fontsize=11)
ax.set_title("Per-Class Precision / Recall / F1 — SVM", fontsize=12)
ax.legend(loc="upper right")
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter("%.2f"))
plt.tight_layout()
plt.show()

print("\nEvaluation complete.")
```
